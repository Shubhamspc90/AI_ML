```python
# 1. What is BERT?
# Answer:

# BERT (Bidirectional Encoder Representations from Transformers) is a Transformer-based language model developed by Google.

# It uses only the Encoder part of the Transformer and reads text in both directions (left and right) to understand context.

# Applications:
# Google Search
# Question Answering
# Sentiment Analysis
# Named Entity Recognition (NER)
# Text Classification

# Interview Answer:
# BERT is a bidirectional Transformer model that uses the Encoder to understand the context of words from both left and right directions.
```

```python
# 2. What is GPT?
# Answer:

# GPT (Generative Pre-trained Transformer) is a Transformer-based language model developed by OpenAI.

# It uses only the Decoder part of the Transformer and generates text one word (or token) at a time.

# Applications:
# Chatbots
# Text Generation
# Code Generation
# Story Writing
# Content Creation

# Interview Answer:
# GPT is a decoder-based Transformer model that generates human-like text by predicting the next word in a sequence.
```

**3. Difference between BERT and GPT?**
BERT is designed for language understanding using the Transformer Encoder, while GPT is designed for text generation using the Transformer Decoder.

| BERT                                          | GPT                                       |
| --------------------------------------------- | ----------------------------------------- |
| Uses Encoder only                             | Uses Decoder only                         |
| Reads text in both directions (Bidirectional) | Reads text left to right (Autoregressive) |
| Best for understanding text                   | Best for generating text                  |
| Used for classification, search, QA           | Used for chatbots, writing, coding        |
| Predicts masked words                         | Predicts the next word                    |


```python
# 4. What is Vision Transformer (ViT)?
# Answer:

# Vision Transformer (ViT) is a Transformer model designed for image processing.

# Instead of using convolution (like CNNs), it:

# Splits an image into small patches.
# Treats each patch like a word (token).
# Processes the patches using the Transformer architecture.
# Applications:
# Image Classification
# Object Detection
# Medical Imaging
# Face Recognition

# Interview Answer:
# Vision Transformer is a Transformer-based model that processes images by dividing them into patches and treating each patch as a token.
```

```python
# 5. Why is GPT Used in ChatGPT?
# Answer:

# GPT is used in ChatGPT because it is designed to:

# Generate natural language
# Continue conversations
# Predict the next word based on context
# Produce coherent and human-like responses

# This makes it well-suited for chat, writing, coding assistance, and question answering.

# Interview Answer:
# GPT is used in ChatGPT because it excels at generating coherent, context-aware, human-like text one token at a time.
```

```python
# 6. Why is BERT Used in Google Search?
# Answer:

# BERT is used in Google Search because it understands the meaning and context of search queries, rather than matching keywords alone.

# Example:

# Query:
# Can you get medicine for someone at a pharmacy?

# BERT understands the relationship between words like "for", "someone", and "pharmacy", helping return more relevant search results.

# Interview Answer:
# BERT is used in Google Search because it understands the context and meaning of search queries, leading to more accurate search results.
```
