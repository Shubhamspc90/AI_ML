```python
# # What is Regularization?

# Regularization is a technique used to improve the generalization of a Deep Learning model.

# Its main purpose is to reduce overfitting and make the model perform well on unseen data.

# Training Data

# ↓

# Model Learning

# ↓

# Regularization

# ↓

# Better Generalization

# ↓

# Accurate Predictions on New Data
```

Import Required Libraries

```python
import tensorflow as tf

from tensorflow import keras

import numpy as np

from tensorflow.keras.callbacks import EarlyStopping
```

```python
# Explanation
# TensorFlow
# ↓
# Build Neural Network

# Keras
# ↓
# Deep Learning API

# EarlyStopping
# ↓
# Stop training automatically when performance stops improving.
```

Create Sample Dataset

```python
X = np.random.rand(

    500,

    4

)

y = np.random.randint(

    0,

    2,

    500

)

print(X.shape)

print(y.shape)
```

ANN with Dropout

```python
model = keras.Sequential([  # Sequential creates a neural network where the layers are connected one after another.

    keras.layers.Dense(

        32,

        activation="relu",

        input_shape=(4,)

    ),

    keras.layers.Dropout(

        0.3

    ),

    keras.layers.Dense(

        16,

        activation="relu"

    ),

    keras.layers.Dense(

        1,

        activation="sigmoid"

    )

])

model.summary()


# Explanation
# Input Layer
# ↓
# Dense Layer
# ↓
# Dropout (30%)
# ↓
# Dense Layer
# ↓
# Output Layer
# Dropout randomly turns off some neurons during training to reduce overfitting.
```

Compile Model

```python
model.compile(
    optimizer = "adam", # Updates weights
    loss = "binary_crossentropy",  # Measures how wrong the prediction is
    metrics =["accuracy"]  # Shows how many predictions are correct
)

# Explanation
# An optimizer is an algorithm that updates the model's weights and biases so that the error becomes smaller.
# Optimizer
# ↓
# Adam -> stands for  (Adaptive Moment Estimation)
# Loss
# ↓
# Binary Cross Entropy
# Metric
# ↓
# Accuracy
```

Early Stopping

```python
early_stop = EarlyStopping( # EarlyStopping automatically stops training when the model is no longer improving. It helps save time and prevent overfitting.

    monitor="loss",

    patience=3,

    restore_best_weights=True

)
```

```python
# Short Explanation

# EarlyStopping automatically stops training when the model is no longer improving. It helps save time and prevent overfitting.

# Line by Line
# 1. monitor="loss"
#   Watches the training loss.
#   If the loss keeps decreasing, training continues.
#   If the loss stops improving, EarlyStopping starts counting.
# 2. patience=3
#   Waits for 3 consecutive epochs with no improvement.
#   If the loss still doesn't improve after 3 epochs, training stops.

# Example:

# Epoch	Loss	Improvement?
# 5	    0.25	✅ Yes
# 6	    0.25	❌ No (Count = 1)
# 7	    0.25	❌ No (Count = 2)
# 8	    0.25	❌ No (Count = 3) → Training Stops

# 3. restore_best_weights=True
#   After training stops, Keras restores the weights from the epoch with the lowest loss.
#   This ensures you keep the best-performing model, not the last one.
```

Train Model

```python
history = model.fit(

    X,

    y,

    epochs=20,

    batch_size=32,

    callbacks=[early_stop],

    verbose=1

)

# Explanation
# Training
# ↓
# Loss Improves
# ↓
# Continue
# Loss Stops Improving
# ↓
# Training Stops Automatically
```

L2 Regularization Example

```python
regularized_model = keras.Sequential([

    keras.layers.Dense(

        32,

        activation="relu",

        kernel_regularizer=keras.regularizers.l2(0.001),

        input_shape=(4,)

    ),

    keras.layers.Dense(

        1,

        activation="sigmoid"

    )

])

regularized_model.summary()

# Explanation

# L2 adds a penalty to large weights.

# This helps the model learn simpler patterns and reduces overfitting.
```

Batch Normalization

```python
batch_norm_model = keras.Sequential([

    keras.layers.Dense(

        32,

        activation="relu",

        input_shape=(4,)

    ),

    keras.layers.BatchNormalization(),

    keras.layers.Dense(

        1,

        activation="sigmoid"

    )

])

batch_norm_model.summary()

# Explanation
# Batch Normalization normalizes the output of a layer.

# Benefits

# Faster Training
# Stable Learning
# Better Performance
```
