```python
# # What is TensorFlow?

# TensorFlow is an open-source Deep Learning framework developed by Google.

# It is used to build, train and deploy Deep Learning models.

# TensorFlow supports

# - CPU
# - GPU
# - TPU

# It is one of the most popular Deep Learning libraries.
```

Import TensorFlow

```python
import tensorflow as tf

print("TensorFlow Version :", tf.__version__)
```

Create Your First Tensor

```python
tensor = tf.constant(10)
print(tensor)

# Explanation
# tf.constant() creates an immutable tensor.
# Once created, its value cannot be changed.
```

Create Multiple Tensors

```python
tensor1 = tf.constant( [1,2,3,4,5] )

tensor2 = tf.constant( [ [1,2],[3,4] ] )

print(tensor1)
print(tensor2)

# Explanation

# 1-D Tensor
# ↓
# Vector

# 2-D Tensor
# ↓
# Matrix
```

Tensor Data Types

```python
int_tensor = tf.constant(

    [1,2,3],

    dtype=tf.int32

)

float_tensor = tf.constant(

    [1.5,2.5,3.5],

    dtype=tf.float32

)

print(int_tensor.dtype)

print(float_tensor.dtype)

# Explanation
# TensorFlow supports many data types.
# Most commonly used
# int32
# float32
# bool
# string
```

Tensor Variables

```python
variable = tf.Variable(10)
print(variable)

variable.assign(20)
print(variable)

# Explanation
# tf.Variable() creates a mutable tensor.
# Variables are updated during model training.
# Weights and Biases are stored as Variables.
```

Tensor Operations

```python

a = tf.constant([1,3,5])
b = tf.constant([2,4,6])

print(tf.add(a,b))

print(tf.subtract(a,b))

print(tf.multiply(a,b))

print(tf.divide(a,b))

# Explanation
# TensorFlow provides built-in mathematical operations.
# Used internally while training Neural Networks.
```

Check GPU Availability

```python
print(

    "GPU Available :",

    tf.config.list_physical_devices("GPU")

)
# Explanation
# Checks whether TensorFlow can access a GPU.
# If the list is empty ([]), TensorFlow is running on CPU.
```

Tensor Shape

```python
tensor = tf.constant(
    [
        [1,2,3],
        [4,5,6]
    ]
)

print(tensor.shape)
print(tf.rank(tensor))

# Explanation
# shape
# ↓
# Number of rows and columns.
# rank
# ↓
# Number of dimensions.
```
