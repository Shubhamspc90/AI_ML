```python
# # What is Conv2D?

# Conv2D extracts image features.

# Examples

# - Edge
# - Curve
# - Eye
# - Nose
# - Ear
# - Tail

# Instead of looking at the complete image,
# CNN learns important features.
```

```python
# # What is MaxPooling?

# Pooling keeps important information and removes unnecessary information.

# Benefits

# - Reduce Image Size
# - Reduce Computation
# - Reduce Overfitting
```

```python
# # Why Flatten?

# CNN Output

# 36 × 36 × 64

# ↓

# Flatten

# ↓

# 82944

# Dense Layers only accept one-dimensional input.
```

```python
# # Why Sigmoid?

# Binary Classification

# Cat

# Dog

# Sigmoid returns a value between

# 0

# ↓

# 1

# If Probability > 0.5

# Dog

# Else

# Cat
```

```python
# # Why EarlyStopping?

# Training should stop automatically when the model stops improving.

# Benefits

# - Saves Time
# - Prevents Overfitting
# - Restores Best Model
```
