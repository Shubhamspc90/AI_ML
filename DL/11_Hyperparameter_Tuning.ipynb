```python
# # What is Hyperparameter Tuning?

# Hyperparameter Tuning is the process of selecting the best values before training the model.

# Good Hyperparameters

# ↓

# Better Training

# ↓

# Higher Accuracy

# ↓

# Better Predictions

# Unlike weights and bias, Hyperparameters are not learned by the model.
```

Import TensorFlow

```python
import tensorflow as tf

from tensorflow import keras

# Explanation

# TensorFlow
# ↓
# Deep Learning Framework

# Keras
# ↓
# Build Neural Network
```

Create Sample Dataset

```python
import numpy as np

X = np.random.rand( 500,4 )
y = np.random.randint(0,2,500)

print(X.shape)
print(y.shape)
```

Learning Rate

```python
optimizer = keras.optimizers.Adam(learning_rate=0.001)
print(optimizer)

# Explanation
# Learning Rate controls how much the weights change after every update.
# Small Learning Rate
# ↓
# Slow Learning
# Large Learning Rate
# ↓
# May overshoot the optimum solution.
```

Number of Neurons

```python
model = keras.Sequential([
        keras.layers.Dense(
            64,
            activation ="relu",
            input_shape=(4,)
        ),
        keras.layers.Dense(
            32,
            activation = "relu"
        ),
        keras.layers.Dense(
            1,
            activation ="sigmoid"
        )
])

model.summary()

# Explanation
# More Neurons
# ↓
# Higher Learning Capacity
# ↓
# May increase Overfitting.
```

Compile Model

```python
model.compile(

    optimizer=optimizer,

    loss="binary_crossentropy",

    metrics=["accuracy"]

)

# Explanation

# Optimizer
# ↓
# Adam

# Loss
# ↓
# Binary Cross Entropy

# Metric
# ↓
# Accuracy
```

Train with Batch Size and Epochs

```python
history = model.fit(
    X,
    y,
    epochs = 10,
    batch_size = 16,
    verbose =1
)

# Explanation

# Epochs
# ↓
# How many times the model sees the complete dataset.

# Batch Size
# ↓
# Number of samples processed before updating weights.
```

Compare Activation Functions

```python
activations = [
    "relu",
    "sigmoid",
    "tanh",
    "softmax"
]

for activation in activations:
    print(activation)
    
# Explanation

# Different activation functions are suitable for different problems.

# Choosing the correct activation function improves model performance.
```

Evaluate Model

```python
loss , accuracy = model.evaluate(
    X,
    y,
    verbose =1
)
print("Loss : ",loss)
print("Accuracy : ",accuracy)

# Explanation
# Evaluate checks the performance of the tuned model after training.
```
