```python
# # What is a Loss Function?

# A Loss Function measures how far the model's prediction is from the actual value.

# Lower Loss = Better Model

# Higher Loss = Poor Model

# Actual Value

# ↓

# Prediction

# ↓

# Loss Function

# ↓

# Error
```

Import Libraries

```python
import tensorflow as tf

import numpy as np
```

Mean Squared Error (MSE)

```python
y_true = tf.constant([10,20,30],dtype=tf.float32)
y_pred = tf.constant([12,18,28],dtype=tf.float32)

mse = tf.keras.losses.MeanSquaredError()
print(mse(y_true , y_pred).numpy())

# (actual - predicted)²
# ↓
# Average  

# Mostly used in

# ✅ Regression
```

Mean Absolute Error (MAE)

```python
mae = tf.keras.losses.MeanAbsoluteError()

print(

    mae(

        y_true,

        y_pred

    ).numpy()

)

# |actual - predicted|
# ↓
# Average

# Used in

# ✅ Regression
```

Binary Cross Entropy

```python

y_true = [1,0,1]

y_pred = [0.95,0.10,0.80]

bce = tf.keras.losses.BinaryCrossentropy()

print(

    bce(

        y_true,

        y_pred

    ).numpy()

)

# Used in

# ✅ Binary Classification

# Examples

# Spam Detection

# Customer Churn

# Disease Prediction
```

Categorical Cross Entropy

```python
y_true = [[0,1,0]]

y_pred = [[0.1,0.8,0.1]]

cce = tf.keras.losses.CategoricalCrossentropy()

print(

    cce(

        y_true,

        y_pred

    ).numpy()

)

# Used in

# ✅ Multi-Class Classification

# Example

# Cat

# Dog

# Horse
```

Sparse Categorical Cross Entropy

```python
import tensorflow as tf

y_true = tf.constant([1], dtype=tf.int32)

y_pred = tf.constant([[0.1, 0.8, 0.1]], dtype=tf.float32)

scce = tf.keras.losses.SparseCategoricalCrossentropy()

print(scce(y_true, y_pred).numpy())
```

Hinge Loss

```python
y_true = [1,-1,1]

y_pred = [0.8,-0.7,0.5]

hinge = tf.keras.losses.Hinge()

print(

    hinge(

        y_true,

        y_pred

    ).numpy()

)

# Mostly used in

# ✅ Support Vector Machine (SVM)

# Rarely used in Deep Learning models.
```

Compare Loss Functions

```python
loss_functions = [

    "MSE",

    "MAE",

    "Binary Cross Entropy",

    "Categorical Cross Entropy",

    "Sparse Categorical Cross Entropy",

    "Hinge"

]

for loss in loss_functions:

    print(loss)
    
# Har problem ke liye alag Loss Function use hoti hai.

# Wrong Loss Function choose karne se model ki performance kharab ho sakti hai.
```
