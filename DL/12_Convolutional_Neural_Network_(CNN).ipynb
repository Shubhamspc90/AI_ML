```python
# # What is CNN?

# A Convolutional Neural Network (CNN) is a Deep Learning model specially designed for image data.

# Instead of looking at the entire image at once, CNN learns small visual patterns like

# - Edges
# - Corners
# - Shapes
# - Objects

# Image
# ↓
# Convolution
# ↓
# Pooling
# ↓
# Flatten
# ↓
# Dense Layer
# ↓
# Prediction
```

Import Required Libraries

```python
import tensorflow as tf
from tensorflow import keras

# Explanation
# TensorFlow
# ↓
# Deep Learning Framework

# Keras
# ↓
# Build CNN Models
```

Load MNIST Dataset

```python
(X_train, y_train), (X_test, y_test) = keras.datasets.mnist.load_data()

print(X_train.shape)

print(X_test.shape)

# Explanation

# MNIST contains
# 60,000 Training Images
# 10,000 Testing Images

# Each image size
# 28 × 28 pixels
```

Normalize Images

```python
X_train = X_train / 255.0

X_test = X_test / 255.0

print(X_train.min(), X_train.max())

# Explanation

# Pixel Values
# 0 → 255
# ↓
# Normalize
# ↓
# 0 → 1
# This helps the model learn faster.
```

Build CNN Model

```python
model = keras.Sequential([

    keras.layers.Conv2D(

        32,

        (3,3),

        activation="relu",

        input_shape=(28,28,1)

    ),

    keras.layers.MaxPooling2D(

        (2,2)

    ),

    keras.layers.Flatten(),

    keras.layers.Dense(

        128,

        activation="relu"

    ),

    keras.layers.Dense(

        10,

        activation="softmax"

    )

])

model.summary()

# Explanation

# Image
# ↓
# Convolution
# ↓
# Pooling
# ↓
# Flatten
# ↓
# Dense
# ↓
# Softmax Output
```

Compile Model

```python
model.compile(

    optimizer="adam",

    loss="sparse_categorical_crossentropy",

    metrics=["accuracy"]

)

# Explanation

# Optimizer
# ↓
# Adam

# Loss
# ↓
# Sparse Categorical Cross Entropy

# Metric
# ↓
# Accuracy
```

Train Model

```python
history = model.fit(

    X_train.reshape(-1,28,28,1),

    y_train,

    epochs=5,

    batch_size=32,

    validation_split=0.2,

    verbose=1

)

# Explanation
# Training Images
# ↓
# CNN Learns Features
# ↓
# Update Weights
# ↓
# Improve Accuracy
```

Evaluate Model

```python
loss, accuracy = model.evaluate(

    X_test.reshape(-1,28,28,1),

    y_test,

    verbose=0

)

print("Accuracy :", accuracy)

# Explanation

# Testing Data
# ↓
# Evaluate Model
# ↓
# Final Accuracy
```

Predict First Image

```python
prediction = model.predict(

    X_test[:1].reshape(-1,28,28,1)

)

print("Predicted Digit :", prediction.argmax())

# Explanation
# CNN
# ↓
# Probability for 10 Digits
# ↓
# Highest Probability
# ↓
# Final Prediction
```

**Quick Interview Revision**

| Question                                  | Short Answer                                                                           |
| ----------------------------------------- | -------------------------------------------------------------------------------------- |
| **What is CNN?**                          | A deep learning model designed for image processing.                                   |
| **Why CNN over ANN?**                     | Preserves image structure, extracts features automatically, and gives better accuracy. |
| **What is Convolution?**                  | Applies filters to extract features like edges and textures.                           |
| **What is Pooling?**                      | Reduces feature map size while keeping important information.                          |
| **Why Flatten?**                          | Converts feature maps into a 1D vector for Dense layers.                               |
| **Why Softmax?**                          | Converts outputs into probabilities for multi-class classification.                    |
| **Why Sparse Categorical Cross Entropy?** | Because MNIST has 10 classes and uses integer labels (0–9).                            |
