```python
# # What is Artificial Neural Network?

# An Artificial Neural Network (ANN) is a Deep Learning model inspired by the human brain.

# It consists of multiple layers of neurons that learn patterns from data.

# Architecture

# Input Layer

# ↓

# Hidden Layer(s)

# ↓

# Output Layer

# Each neuron receives information, processes it, and passes it to the next layer.
```

Import Required Libraries

```python
import tensorflow as tf

from tensorflow import keras

import numpy as np

# Explanation
# TensorFlow
# ↓
# Deep Learning Framework

# Keras
# ↓
# Build Neural Networks

# NumPy
# ↓
# Create Sample Dataset
```

Create Dummy Dataset

```python
X = np.random.rand(
    500,
    4
)

y = np.random.randint(
    0,
    2,
    500
)

print(X.shape)
print(y.shape)

# Explanation
# 500 Samples
# ↓
# 4 Input Features
# ↓
# Binary Target (0 or 1)
```

Build First ANN

```python
model = keras.Sequential([

    keras.layers.Dense(

        16,

        activation="relu",

        input_shape=(4,)

    ),

    keras.layers.Dense(

        8,

        activation="relu"

    ),

    keras.layers.Dense(

        1,

        activation="sigmoid"

    )

])

model.summary()


# Explanation
# Input Layer
# ↓
# 16 Neurons
# ↓
# 8 Neurons
# ↓
# 1 Output Neuron
```

Compile Model

```python
model.compile(

    optimizer="adam",

    loss="binary_crossentropy",

    metrics=["accuracy"]

)

# Explanation
# Optimizer
# ↓
# Adam
# Loss
# ↓
# Binary Cross Entropy
# Metric
# ↓
# Accuracy
```

Train Model

```python
history = model.fit(

    X,

    y,

    epochs=10,

    batch_size=32,

    verbose=1

)

# Explanation
# Training Data
# ↓
# Forward Propagation
# ↓
# Loss Calculation
# ↓
# Backpropagation
# ↓
# Weight Update
# ↓
# Repeat for Every Epoch
```

Evaluate Model

```python
loss, accuracy = model.evaluate(
    X,
    y,
    verbose=0
)

print("Loss :", loss)
print("Accuracy :", accuracy)

# Explanation
# Evaluate tells us how well the model performs on the given data.
```

Make Predictions

```python
prediction = model.predict(
    X[:5]
)
print(prediction)

# Explanation
# The output is a probability between 0 and 1.
# Example
# 0.92 → Class 1
# 0.08 → Class 0
```

Convert Probability into Class

```python
predicted_class = (

    prediction > 0.5

).astype(int)

print(predicted_class)

# Explanation
# Probability
# ↓
# Threshold (0.5)
# ↓
# Final Prediction
# 0
# or
# 1
```
