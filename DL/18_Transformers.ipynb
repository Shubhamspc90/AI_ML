```python
# # Why Transformers?

# RNN reads one word at a time.

# ↓

# LSTM remembers long-term information.

# ↓

# But both process words sequentially.

# ↓

# Transformers process all words together using Self-Attention.

# This makes training much faster and more accurate.
```

Input Sentence

```python
sentence = [ "I", "love", "Deep", "Learning" ]
print(sentence)

# Explanation
# This is our input sequence.
# Unlike RNN,
# Transformer sees the whole sentence together.
```

Sequential vs Parallel Processing

```python
rnn = [ "I", "→", "love", "→", "Deep", "→", "Learning" ]

transformer = [ "I", "love", "Deep", "Learning" ]

print("RNN :", rnn)

print()

print("Transformer :", transformer)

# Explanation

# RNN

# Word
# ↓
# Next Word
# ↓
# Next Word

# Transformer

# All Words
# ↓
# Together
# ↓
# Prediction
```

Positional Encoding

```python
positions = {
    "I":0,
    "love":1,
    "Deep":2,
    "Learning":3
}

print(positions)

# Explanation
# Since Transformer reads all words together,
# it needs to know each word's position.
# Positional Encoding provides this information.
```

Self-Attention

```python
attention = {

    "Learning":["Deep","love"],

    "love":["I"],

    "Deep":["Learning"]

}

print(attention)

# Explanation
# Every word checks
# which other words are important.
```

Quick Interview Revision

| Question                                       | Short Answer                                                                                                         |
| ---------------------------------------------- | -------------------------------------------------------------------------------------------------------------------- |
| **Why are Transformers faster than RNN?**      | They process all words in parallel instead of one by one.                                                            |
| **What is Positional Encoding?**               | It adds word-order information so the Transformer knows the position of each word.                                   |
| **What is the Encoder?**                       | It converts the input sequence into context-aware representations.                                                   |
| **What is the Decoder?**                       | It generates the output sequence using the encoder's output.                                                         |
| **Why is Self-Attention important?**           | It lets every word attend to all other words, improving context understanding and long-range dependency learning.    |
| **Why did Transformers replace LSTMs in NLP?** | They are faster, handle long-range dependencies better, scale efficiently, and generally achieve better performance. |
