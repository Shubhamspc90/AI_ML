```python
# # What is Keras?

# Keras is TensorFlow's high-level API used to build Deep Learning models easily.

# With Keras we can

# - Build Models
# - Train Models
# - Evaluate Models
# - Make Predictions
# - Save Models

# Keras reduces the amount of code required to build Neural Networks.
```

Import Required Libraries

```python
import tensorflow as tf
from tensorflow import keras

print(tf.__version__)
print(keras.__version__)

model = keras.Sequential()

print(model)

# Explanation
# TensorFlow
# ↓
# Deep Learning Framework
# Keras
# ↓
# High-Level API for building Neural Networks
```

Create First Model using Sequential API


```python
model = keras.Sequential()
print(model)
```

```python
# Explanation
# Sequential API is the easiest way to build a Neural Network.
# Layers are added one after another.
```

Add Dense Layers

```python
model = keras.Sequential([
    keras.layers.Dense(
        16,  # This layer contains 16 neurons.
        activation="relu",  # Rectified Linear Unit , formula=> f(x)=max(0,x)
        input_shape=(4,)
    ),

    keras.layers.Dense(
        8,
        activation="relu"
    ),

    keras.layers.Dense(
        1,
        activation="sigmoid"
    )
])

model.summary()

# Explanation
# Input Layer
# ↓
# Hidden Layer (16 Neurons)
# ↓
# Hidden Layer (8 Neurons)
# ↓
# Output Layer (1 Neuron)
```

Compile Model

```python
model.compile(

    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]

)

print("Model Compiled Successfully")

# Explanation
# Optimizer
# ↓
# Adam
# Loss
# ↓
# Binary Cross Entropy
# Metric
# ↓
# Accuracy
```

Create Dummy Dataset

```python
import numpy as np

X = np.random.rand(100,4)

y = np.random.randint(
    0,
    2,
    100
)

print(X.shape)

print(y.shape)
```

Train Model

```python
history = model.fit(  # The fit() method is used to train the neural network.

    X,
    y,  # This is the target (label).
    epochs=5, # the model repeats this process 5 times., One complete pass = 1 Epoch
    batch_size=16,
    verbose=1  # This controls how much information is displayed during training.
   # verbose=0 => No Output
)

# Explanation
# Model learns from the training data.
# Weights are updated after every batch.
```

```python
import matplotlib.pyplot as plt

plt.plot(history.history["loss"])
plt.title("Training Loss")
plt.show()
```

```python
plt.plot(history.history["accuracy"])
plt.title("Training Accuracy")
plt.show()
```

Evaluate & Predict

```python
loss, accuracy = model.evaluate(

    X,

    y,

    verbose=0

)

print("Accuracy :", accuracy)

prediction = model.predict(

    X[:3]

)

print(prediction)

# Explanation
# Evaluate
# ↓
# Model Performance
# Predict
# ↓
# Output Probability
```

Save & Load Model

```python
model.save(

    "keras_model.keras"

)

loaded_model = keras.models.load_model(

    "keras_model.keras"

)

print("Model Loaded Successfully")

# Explanation
# Model
# ↓
# Save
# ↓
# Load
# ↓
# Use Again
# No need to retrain the model.
```
