Import TensorFlow

```python
import tensorflow as tf

from tensorflow import keras

# Why Import?
# TensorFlow and Keras provide built-in NLP datasets and preprocessing utilities for Deep Learning.
```

Load IMDb Dataset

```python
(X_train, y_train), (X_test, y_test) = keras.datasets.imdb.load_data(

    num_words=10000

)

print("Training Reviews :", len(X_train))

print("Testing Reviews :", len(X_test))

# Explanation
# IMDb Dataset
# ↓
# 50,000 Movie Reviews
# ↓
# Positive / Negative Labels
# ↓
# Ready for NLP
```

Explore One Review

```python
print("Encoded Review:")

print(X_train[0])

print()

print("Sentiment Label:")

print(y_train[0])

# Explanation

# Notice that the review is not stored as text.

# Each word has been converted into an integer.

# Example

# the → 1

# movie → 17

# good → 49

# Computers understand numbers better than raw text.
```

Load Word Index

```python
word_index = keras.datasets.imdb.get_word_index()

print(list(word_index.items())[:10])

# Why?
# Word Index tells us
# Word
# ↓
# Integer Number
# Example
# good → 49
# movie → 17
```

Decode an Encoded Review

```python
reverse_word_index = {

    value + 3: key

    for key, value in word_index.items()

}

reverse_word_index[0] = "<PAD>"

reverse_word_index[1] = "<START>"

reverse_word_index[2] = "<UNK>"

reverse_word_index[3] = "<UNUSED>"

decoded_review = " ".join(

    reverse_word_index.get(

        word,

        "?"

    )

    for word in X_train[0]

)

print(decoded_review)

# Explanation
# Numbers
# ↓
# Words
# ↓
# Original Movie Review
# Now we can actually read what the review says.
```

Apply Padding

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences

X_train = pad_sequences(

    X_train,

    maxlen=200

)

X_test = pad_sequences(

    X_test,

    maxlen=200

)

print(X_train.shape)

# Why Import?
# pad_sequences() makes every review the same length.
# 💡 Flow
# Different Length Reviews
# ↓
# Padding
# ↓
# Same Length
# ↓
# Ready for Deep Learning
```

Check First Padded Review

```python
print(X_train[0])

print()

print("Length :", len(X_train[0]))
```

Explanation

Every review now contains exactly 200 words, making it suitable for neural networks.

Complete NLP Pipeline

```python
steps = [

    "Raw Text",

    "Integer Encoding",

    "Padding",

    "Embedding",

    "LSTM / GRU"

]

for step in steps:

    print(step)
    
# Explanation
# Movie Review
# ↓
# Integer Encoding
# ↓
# Padding
# ↓
# Embedding Layer
# ↓
# LSTM / GRU
# ↓
# Prediction
```

Mini Practical

```python
review_number = 5

print("Review Label :", y_train[review_number])

decoded = " ".join(

    reverse_word_index.get(

        word,

        "?"

    )

    for word in X_train[review_number]

)

print()

print(decoded)

# Explanation
# Pick any review.
# ↓
# Decode it.
# ↓
# Read it.
# ↓
# Compare with the sentiment label.
```

Quick Interview Revision

| Question                                                | Short Answer                                                                                             |
| ------------------------------------------------------- | -------------------------------------------------------------------------------------------------------- |
| **What is NLP?**                                        | A branch of AI that enables computers to understand and process human language.                          |
| **Why can't Deep Learning models understand raw text?** | Because they only work with numerical data, not words.                                                   |
| **What is Integer Encoding?**                           | Converting each word into a unique integer ID.                                                           |
| **What is Word Index?**                                 | A dictionary that maps words to integer IDs.                                                             |
| **Why do we use `pad_sequences()`?**                    | To make all text sequences the same length.                                                              |
| **What is the purpose of decoding a review?**           | To convert integer sequences back into readable text.                                                    |
| **Why is padding required before using LSTM or GRU?**   | Because all sequences in a batch must have the same length for these models to process them efficiently. |
