Artificial Neuron

Artificial Neuron biological neuron ka mathematical version hai.

It receives multiple inputs.

Each input has a weight.

Bias add hota hai.

Activation Function final output decide karti hai.

Artificial Neuron

```python
# x1 -----

#           \

# x2 ------> Σ ----> Activation ----> Output

#           /

# x3 -----

#         +

#        Bias
```

Create First Artificial Neuron

```python

inputs = np.array([2,3])

weights = np.array([0.5,0.8])

bias = 1

output = np.dot(

    inputs,

    weights

) + bias

print(output)
```

```python
# Explanation

# Formula

# Output = (Inputs × Weights) + Bias

# Ye neuron ka basic calculation hai.
```

Add Activation Function

```python
relu_output = tf.nn.relu(

    output

)

print(relu_output.numpy())
```

What is Perceptron?

Perceptron is the first Artificial Neural Network.

It contains

Inputs
Weights
Bias
Activation Function
Output

Ye sirf Single Layer Network hota hai.

Perceptron Example

```python
# Create the model
model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(2,)),  # Input layer
    tf.keras.layers.Dense(
        units=1,
        activation="relu"
    )
])

# Display model summary
model.summary()
```

Multi Layer Perceptron (MLP)

```python
mlp = tf.keras.Sequential([

    tf.keras.layers.Dense(8,activation="relu"),

    tf.keras.layers.Dense(4,activation="relu"),

    tf.keras.layers.Dense(1,activation="sigmoid")

])

mlp.summary()
```

Forward Propagation

```python
sample = np.array([[2,3]])

prediction = mlp(

    sample

)

prediction
```
