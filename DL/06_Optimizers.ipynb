```python
# # What is an Optimizer?

# An Optimizer updates the model's weights and bias to reduce the Loss Function.
# Loss
# ↓
# Optimizer
# ↓
# Update Weights
# ↓
# Better Prediction

# The main goal of an optimizer is to minimize the loss and improve model accuracy.
```

Import TensorFlow

```python
import tensorflow as tf

print("TensorFlow Version :", tf.__version__)
```

Gradient Descent

```python
optimizer = tf.keras.optimizers.SGD(

    learning_rate=0.01

)

print(optimizer)

# Explanation

# Gradient Descent is the basic optimizer.
# It updates the weights in the direction where the loss decreases.
```

Stochastic Gradient Descent (SGD)

```python
sgd = tf.keras.optimizers.SGD(
    learning_rate = 0.01,
    momentum = 0.9
)
print(sgd)

# Explanation
# SGD updates weights frequently.
# Momentum helps the optimizer move faster and reduces oscillations.

# Used in
# ✔ CNN
# ✔ Image Classification
```

RMSProp

```python
rmsprop = tf.keras.optimizers.RMSprop(

    learning_rate=0.001

)

print(rmsprop)

# Explanation

# RMSProp automatically adjusts the learning rate during training.
# Works well for sequence data.
```

Adam Optimizer

```python
adam = tf.keras.optimizers.Adam(

    learning_rate=0.001

)

print(adam)

# Explanation
# Adam combines
# Momentum
# Adaptive Learning Rate
# Most commonly used optimizer in Deep Learning.
```

AdamW Optimizer

```python
adamw = tf.keras.optimizers.AdamW(

    learning_rate=0.001,

    weight_decay=0.0001

)

print(adamw)

# Explanation
# AdamW is an improved version of Adam.
# It applies Weight Decay to reduce overfitting.
# Mostly used in
# ✔ Transformers
# ✔ LLMs
# ✔ Modern Deep Learning Models
```

Use Optimizer in a Neural Network

```python
model = tf.keras.Sequential([

    tf.keras.layers.Dense(

        8,

        activation="relu",

        input_shape=(4,)

    ),

    tf.keras.layers.Dense(

        1,

        activation="sigmoid"

    )

])

model.compile(

    optimizer="adam",

    loss="binary_crossentropy",

    metrics=["accuracy"]

)

print("Model Compiled Successfully")

# Explanation
# Optimizer
# ↓
# Adam
# Loss
# ↓
# Binary Cross Entropy
# Metric
# ↓
# Accuracy
```

Compare Optimizers

```python
optimizers = [

    "Gradient Descent",

    "SGD",

    "RMSProp",

    "Adam",

    "AdamW"

]

for optimizer in optimizers:

    print(optimizer)
    
    
# Different optimizers work better for different problems.
# Adam is the default choice for most Deep Learning projects.
```
