```python
# # Why Attention?

# RNN reads words one by one.

# LSTM remembers important information for a long time.

# But for very long sentences, even LSTM struggles to remember everything.

# Attention allows the model to focus on the most important words while making predictions.

# Sentence

# ↓

# Attention

# ↓

# Important Words

# ↓

# Prediction
```

A Long Sentence

```python
sentence = [

    "The",

    "movie",

    "was",

    "very",

    "interesting",

    "and",

    "the",

    "acting",

    "was",

    "excellent"

]

print(sentence)

# Explanation
# This is a sequence of words.
# If we want to predict the sentiment,
# not every word is equally important.
```

Simulate Attention

```python
attention_scores = {

    "The":0.05,

    "movie":0.20,

    "interesting":0.80,

    "acting":0.90,

    "excellent":1.00

}

print(attention_scores)

# Explanation
# Attention gives more importance to meaningful words.
# Example
# excellent → High Attention
# acting → High Attention
# The → Low Attention
```

Sort Important Words

```python
important_words = sorted(

    attention_scores.items(),

    key=lambda x:x[1],

    reverse=True

)

print(important_words) 

# Explanation
# Now we can clearly see
# which words influence the prediction the most.
```

Self-Attention Example

```python
words = [ "I", "love", "Deep", "Learning" ]

for word in words:
    print( f"{word} attends to other words." )
    
# Explanation
# In Self-Attention,
# every word looks at every other word before making its representation.
```

Multi-Head Attention Concept

```python
heads = [ "Grammar", "Meaning", "Context" ]

for head in heads:
    print(head)
    
# Explanation
# Different attention heads learn different relationships.
# One head may learn grammar.
# Another may learn meaning.
# Another may learn context.
```

**Difference between LSTM and Attention?**
| LSTM                                   | Attention                               |
| -------------------------------------- | --------------------------------------- |
| Processes words one by one             | Processes all words together            |
| Sequential processing                  | Parallel processing                     |
| Slower                                 | Faster                                  |
| Harder to learn very long dependencies | Easily captures long-range dependencies |
| Has memory cells                       | Uses attention scores                   |


**Quick Interview Revision**
| Question                                         | Short Answer                                                                                                                      |
| ------------------------------------------------ | --------------------------------------------------------------------------------------------------------------------------------- |
| **Why was Attention introduced?**                | To focus on important words and improve learning of long-range dependencies.                                                      |
| **What problem does Attention solve?**           | It solves the long-term dependency problem by allowing direct access to important words.                                          |
| **What is Self-Attention?**                      | Each word attends to every other word in the same sequence to understand context.                                                 |
| **What is Multi-Head Attention?**                | Multiple attention heads learn different relationships in parallel.                                                               |
| **Difference between LSTM and Attention?**       | LSTM is sequential; Attention is parallel and handles long dependencies more effectively.                                         |
| **Why is Attention important for Transformers?** | It is the core mechanism that enables Transformers to process sequences in parallel and achieve state-of-the-art NLP performance. |
