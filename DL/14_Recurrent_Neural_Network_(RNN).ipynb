```python
# # What is Recurrent Neural Network (RNN)?

# A Recurrent Neural Network (RNN) is a Deep Learning model designed for sequential data.

# Unlike ANN, an RNN remembers information from previous time steps while processing the current input.

# Movie Review

# ↓

# Word Sequence

# ↓

# RNN

# ↓

# Learn Context

# ↓

# Positive / Negative Prediction

# RNN is widely used in

# - Sentiment Analysis
# - Language Translation
# - Speech Recognition
# - Time Series Forecasting
```

Import TensorFlow

```python
import tensorflow as tf

from tensorflow import keras

# Explanation
# TensorFlow
# ↓
# Deep Learning Framework
# Keras
# ↓
# Build RNN Models
```

Load IMDb Dataset

```python
(X_train, y_train), (X_test, y_test) = keras.datasets.imdb.load_data(

    num_words=10000

)

print(X_train.shape)

print(X_test.shape)

# Explanation

# IMDb Dataset

# ↓

# 50,000 Movie Reviews

# ↓

# Top 10,000 Most Frequent Words

# ↓

# Binary Labels

# Positive

# or

# Negative
```

Pad Sequences

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences

X_train = pad_sequences(

    X_train,

    maxlen=200

)

X_test = pad_sequences(

    X_test,

    maxlen=200

)

print(X_train.shape)

# Explanation

# Movie Reviews

# ↓

# Different Lengths

# ↓

# Same Length (200 Words)

# ↓

# Ready for RNN
```

Build First RNN

```python
model = keras.Sequential([

    keras.layers.Embedding(

        input_dim=10000,

        output_dim=64,

        input_length=200

    ),

    keras.layers.SimpleRNN(

        64

    ),

    keras.layers.Dense(

        1,

        activation="sigmoid"

    )

])

model.summary()

# Explanation

# Word IDs

# ↓

# Embedding Layer

# ↓

# SimpleRNN

# ↓

# Dense Layer

# ↓

# Prediction
```

Compile Model

```python
model.compile(
    optimizer = "adam",
    loss = "binary_crossentropy",
    metrics=["accuracy"]
)

# Explanation

# Optimizer
# ↓
# Adam

# Loss
# ↓
# Binary Cross Entropy

# Metric
# ↓
# Accuracy
```

Train Model

```python
history = model.fit(
    X_train,
    y_train,
    epochs=3,
    batch_size=64,
    validation_split=0.2,
    verbose=1
)

# Explanation

# Movie Reviews

# ↓

# Embedding

# ↓

# RNN Learns Context

# ↓

# Weight Update

# ↓

# Better Accuracy
```

Evaluate Model

```python
loss, accuracy = model.evaluate(

    X_test,

    y_test,

    verbose=0

)

print("Accuracy :", accuracy)

# Explanation
# Evaluate
# ↓
# Check Model Performance
# ↓
# Final Accuracy
```

Mini Project – Predict Review Sentiment

```python
prediction = model.predict(

    X_test[:1]

)

print("Prediction Probability :", prediction[0][0])

if prediction[0][0] > 0.5:

    print("Positive Review")

else:

    print("Negative Review")
    
# Explanation

# New Movie Review

# ↓

# Embedding

# ↓

# RNN

# ↓

# Positive / Negative
```
