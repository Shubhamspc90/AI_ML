```python
# # What is an Activation Function?

# Activation Function decides whether a neuron should be activated or not.

# Without an activation function, a Neural Network behaves like a simple linear model and cannot learn complex patterns.

# Activation Function

# ↓

# Input

# ↓

# Calculation

# ↓

# Output
```

Import Libraries

```python
import tensorflow as tf
import  numpy as np
import matplotlib.pyplot as plt

```

ReLU Activation

# Formula
# ReLU(x)=max(0,x)

# This means:

# If x > 0, output = x
# If x ≤ 0, output = 0

```python
x = tf.constant([-5,-2,0,2,5],dtype=tf.float32)

relu = tf.nn.relu(x)

print(relu.numpy())

x = tf.constant([-5,-2,0,2,5],dtype=tf.float32)

relu = tf.nn.relu(x)

# Explanation

# Negative Values → 0
# Positive Values → Same
# Formula
# f(x)=max(0,x)
# ReLU is the most commonly used activation function in hidden layers.
```

Sigmoid Activation

```python
sigmoid = tf.nn.sigmoid(x)

print(sigmoid.numpy())

# Explanation
# Output Range
# 0 → 1
# Mostly used for
# Binary Classification
# Formula
# 1 / (1 + e⁻ˣ)
```

Tanh Activation

```python
tanh = tf.nn.tanh(x)

print(tanh.numpy())

# Explanation
# Output Range
# -1 → 1
# Zero-centered activation.
# Usually performs better than Sigmoid in many cases.
```

Leaky ReLU

```python
leaky_relu = tf.nn.leaky_relu(

    x,

    alpha=0.1

)

print(leaky_relu.numpy())

# Explanation
# Negative values completely zero nahi hote.
# Small slope maintain hoti hai.
# Dead ReLU problem ko reduce karta hai.
```

Softmax Activation

```python
values = tf.constant(

    [2.0,4.0,6.0]

)

softmax = tf.nn.softmax(values)

print(softmax.numpy())

# Explanation
# Output
# Probability Distribution 
# Sab values ka sum = 1

# Used in
# Multi-Class Classification
```

Compare Activation Functions

```python
x = np.linspace(-5,5,100)

plt.figure(figsize=(8,5))

plt.plot(x,tf.nn.sigmoid(x),label="Sigmoid")

plt.plot(x,tf.nn.tanh(x),label="Tanh")

plt.plot(x,tf.nn.relu(x),label="ReLU")

plt.legend()

plt.grid()

plt.show()
```

<!-- Explanation

Graph clearly shows

Sigmoid → Smooth curve (0 to 1)
Tanh → Smooth curve (-1 to 1)
ReLU → Zero for negatives, linear for positives -->

```python
# Explanation

# Graph clearly shows

# Sigmoid → Smooth curve (0 to 1)
# Tanh → Smooth curve (-1 to 1)
# ReLU → Zero for negatives, linear for positives
```

Using Activation Function in Neural Network

```python
model = tf.keras.Sequential([
    tf.keras.layers.Dense( 8,  activation="relu"),

    tf.keras.layers.Dense( 1, activation="sigmoid")
])

model.summary()

# Explanation
# Hidden Layer
# ↓
# ReLU
# Output Layer
# ↓
# Sigmoid
# Ye Binary Classification ka standard ANN architecture hai.
```

When to Use Which?

| Activation | Use Case                   |
| ---------- | -------------------------- |
| ReLU       | Hidden Layers              |
| Leaky ReLU | Dead ReLU Problem          |
| Sigmoid    | Binary Classification      |
| Softmax    | Multi-Class Classification |
| Tanh       | Zero-Centered Data         |
| Step       | Learning Purpose Only      |
