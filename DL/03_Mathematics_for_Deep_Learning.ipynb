Import Libraries

```python
import numpy as np

import tensorflow as tf
```

**Scalar**

Scalar ek single numerical value hoti hai.

Examples

5

10

3.14

Temperature

Age

Scalar Example

```python
scalar = 10
print(scalar)
print(type(scalar))

# Explanation
# Scalar sirf ek value hoti hai.
# Deep Learning me learning rate, bias etc. scalar hote hain.
```

**Vector**

Vector ek 1-D collection hota hai.

Example

[2,4,6]

**Vector Example**

```python
vector = np.array(
    [2,4,6]
)
print(vector)
print(vector.shape)

# Explanation
# Vector multiple values ko ek line me store karta hai.
# Deep Learning me ek image ka row ya features vector ho sakte hain.
```

**Matrix**

Matrix rows aur columns ka collection hota hai.

**Matrix Example**

```python
matrix = np.array(
    [
        [1,2],
        [3,4]
    ]
)

print(matrix)
print(matrix.shape)
```

**Tensor**

Tensor multi-dimensional data hota hai.

TensorFlow isi wajah se TensorFlow kehlata hai.

**Tensor Example**

```python
tensor = tf.constant(
    [
        [1,2],
        [3,4]
    ]
)

print(tensor)
```

**Dot Product**

```python
vector1 = np.array( [2,3] )

vector2 = np.array( [4,5])

dot = np.dot( vector1,vector2)

print(dot)

# Neural Network me har neuron Dot Product karta hai.
```

**Matrix Multiplication**

```python
A = np.array( [ [1,2],[3,4] ] )

B = np.array( [ [5,6],[7,8] ] )

result = np.matmul(A,B)

print(result)
# Explanation
# Deep Learning me har layer Matrix Multiplication karti hai.
# Ye sabse important mathematical operation hai.
```

Tensor Operations

```python
tensor1 = tf.constant([1,2,3])
tensor2 = tf.constant([4,5,6])

print(tf.add(tensor1,tensor2))
```
