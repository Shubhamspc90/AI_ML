```python
# # Why Do We Need LSTM?

# A Simple RNN struggles to remember information for long sequences.

# As the sequence becomes longer, important information gradually disappears.

# This problem is called the Vanishing Gradient Problem.

# Sentence

# ↓

# Simple RNN

# ↓

# Information Lost

# ↓

# Wrong Prediction

# LSTM was introduced to solve this problem.

# It remembers important information for a longer period of time.
```

Import Required Libraries

```python
import tensorflow as tf

from tensorflow import keras

# Explanation

# TensorFlow
# ↓
# Deep Learning Framework

# ↓

# Keras
# ↓
# LSTM & GRU Layers
```

Load IMDb Dataset

```python
(X_train, y_train), (X_test, y_test) = keras.datasets.imdb.load_data(

    num_words=10000

)

from tensorflow.keras.preprocessing.sequence import pad_sequences

X_train = pad_sequences(

    X_train,

    maxlen=200

)

X_test = pad_sequences(

    X_test,

    maxlen=200
)

# Explanation

# IMDb Reviews
# ↓
# Convert to Equal Length
# ↓
# Ready for LSTM
```

```python
# # What are LSTM Gates?

# LSTM uses three gates to control information flow.

# Forget Gate

# ↓

# Input Gate

# ↓

# Output Gate

# These gates help the model remember useful information and forget unnecessary information.
```

Build LSTM Model

```python
model = keras.Sequential([

    keras.layers.Embedding(

        input_dim=10000, # The model knows 10,000 unique words.

        output_dim=64,  # Each word is represented using 64 numbers.

        input_length=200  # Every input sentence has 200 words.

    ),

    keras.layers.LSTM(

        64

    ),

    keras.layers.Dense(

        1,

        activation="sigmoid"

    )

])

model.summary()

# Explanation

# Word IDs
# ↓
# Embedding
# ↓
# LSTM
# ↓
# Dense Layer
# ↓
# Positive / Negative
```

Compile Model

```python
model.compile(

    optimizer="adam",

    loss="binary_crossentropy",

    metrics=["accuracy"]

)
```

Train Model

```python
history = model.fit(

    X_train,

    y_train,

    epochs=3,

    batch_size=64,

    validation_split=0.2,

    verbose=1

)

# Explanation

# Movie Reviews
# ↓
# Embedding
# ↓
# LSTM Learns Long-Term Context
# ↓
# Sentiment Prediction
```

Evaluate Model

```python
loss, accuracy = model.evaluate(

    X_test,

    y_test,

    verbose=0

)

print("Accuracy :", accuracy)
```

Build GRU Model

```python
gru_model = keras.Sequential([

    keras.layers.Embedding(

        input_dim=10000,

        output_dim=64,

        input_length=200

    ),

    keras.layers.GRU(

        64

    ),

    keras.layers.Dense(

        1,

        activation="sigmoid"

    )

])

gru_model.summary()

# Explanation
# Embedding
# ↓
# GRU
# ↓
# Dense Layer
# ↓
# Prediction
```

Compile GRU Model

```python
gru_model.compile(

    optimizer="adam",

    loss="binary_crossentropy",

    metrics=["accuracy"]

)
```

train GRU Model

```python
gru_history = gru_model.fit(

    X_train,

    y_train,

    epochs=3,

    batch_size=64,

    validation_split=0.2,

    verbose=1

)
```

Evaluate GRU Model

```python
loss, accuracy = gru_model.evaluate(

    X_test,

    y_test,

    verbose=0

)

print("GRU Accuracy :", accuracy)
```

Predict using GRU

```python
prediction = gru_model.predict(

    X_test[:1]

)

print(prediction)

if prediction[0][0] > 0.5:

    print("Positive Review")

else:

    print("Negative Review")
```

LSTM vs GRU Performance

| Feature          | LSTM                      | GRU                      |
| ---------------- | ------------------------- | ------------------------ |

| Training Speed   | Slower                    | Faster                   |
| Memory           | Better for Long Sequences | Good                     |

| Best Use Case    | Long Documents            | Short & Medium Sequences |

**Quick Interview Revision**

| Question                                    | Short Answer                                                                                                                       |
| ------------------------------------------- | ---------------------------------------------------------------------------------------------------------------------------------- |
| **Why was LSTM introduced?**                | To solve the vanishing gradient problem and learn long-term dependencies.                                                          |
| **What is the Vanishing Gradient Problem?** | Gradients become very small during training, making it difficult to learn long-term relationships.                                 |
| **What are the three gates in LSTM?**       | Forget Gate, Input Gate, and Output Gate.                                                                                          |
| **Difference between LSTM and GRU?**        | LSTM has 3 gates and is more complex; GRU has 2 gates and is simpler and faster.                                                   |
| **Why is GRU faster than LSTM?**            | It has fewer gates, no separate cell state, and fewer parameters.                                                                  |
| **When should you choose LSTM over GRU?**   | For large datasets and tasks requiring long-term memory; choose GRU when faster training and lower computation are more important. |
